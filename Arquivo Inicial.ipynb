### Importando o pandas

```python
import pandas as pd
pd.options.display.float_format = '{:.2f}'.format
```

### Criando um dataframe a partir de um dicionário

```python
# dataframe = pd.DataFrame()
venda = {'data': ['15/02/2021', '16/02/2021'],
         'valor': [500, 300],
         'produto': ['feijao', 'arroz'],
         'qtde': [50, 70],
        }
vendas_df = pd.DataFrame(venda)
```

### Adicionar 1 linha
    - Linhas de um complemento da base de dados

```python
vendas_dez_df = pd.read_excel("Vendas - Dez.xlsx")

vendas_df = pd.concat([vendas_df, vendas_dez_df], ignore_index = True)
display(vendas_df)
```

### Excluir linhas e colunas

```python
vendas_df = vendas_df.drop("Juros", axis=1)
display(vendas_df)
#eixo 0 é a linha, eixo 1 é a coluna, axis = 0
```

### Valores Vazios
    - Deletar linhas/colunas vazias
    - Deletar linhas que possuem valores vazios
    - Preencher valores vazios (média e último valor)

```python
#deletar linhas e colunas completamente vazias
#vendas_df = vendas_df.dropna(how="all", axis=1)
#display(vendas_df)

#deletar linhas e colunas com qualquer linha ou coluna vazias
#vendas_df = vendas_df.dropna()
#display(vendas_df)

#preencher valores vazios com a média ou último valor
#preencher com a média da coluna=>
vendas_df["Valor Final"] = vendas_df["Valor Final"].fillna(vendas_df["Valor Final"].mean)
display(vendas_df)
```
